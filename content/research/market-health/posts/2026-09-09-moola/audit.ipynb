# Moola: pool events, actual oracle quotes and loans

This notebook reproduces the bundled historical evidence without network access. Selection is exploratory; credit totals are gross tokens, not profit or losses. The two arithmetic implementations share one indexer snapshot.

```python
import json, subprocess, sys
from pathlib import Path
ROOT = Path.cwd()
def checked(script):
    result = subprocess.run([sys.executable, script], cwd=ROOT, text=True, capture_output=True)
    print(result.stdout)
    if result.returncode:
        print(result.stderr, file=sys.stderr)
    result.check_returncode()
checked("analyze.py")
```

```python
checked("independent_audit.py")
```
